This notebook performs cleaning decisions documented in 01_data_understanding.ipynb. No decisions are made here that weren't planned yesterday.

```python
import pandas as pd

df = pd.read_excel("../data/raw/Online Retail.xlsx")
```

```python
df.shape
df.isnull().sum()
```

```python
df["IsCancelled"] = df["InvoiceNo"].astype(str).str.startswith('c')
```

```python
df.head(20)
```

```python
df["CustomerID"] = df["CustomerID"].astype(str)
df["CustomerID"] = df["CustomerID"].replace("nan", "Null")
```

```python
df.info()
```

```python
print(df["CustomerID"].dtype)
```

```python
df["Description"] = df["Description"].fillna("Missing_Description")
```

```python
negative_aty_df = df[df["Quantity"] < 0]
print(negative_aty_df["IsCancelled"].value_counts(dropna = False))
```

```python
are_all_cancelled = negative_aty_df["IsCancelled"].all()
print(are_all_cancelled)
```

```python
invalid_price = df[df["UnitPrice"] <= 0]
print(invalid_price.shape)
```

```python
df['InvoiceDate'] = pd.to_datetime(df["InvoiceDate"])
df["Year"] = df["InvoiceDate"].dt.year
df["Month"] = df["InvoiceDate"].dt.month
df["Day"] = df["InvoiceDate"].dt.day
df["Hour"] = df["InvoiceDate"].dt.hour
df["DayOfWeek"] = df["InvoiceDate"].dt.day_name()
```

```python
# Revenue 

df["Revenue"] = df['Quantity'] * df['UnitPrice']
```

```python
df.to_csv("../data/processed/online_retail_clean.csv", index= False)

print("df was saved")
```
