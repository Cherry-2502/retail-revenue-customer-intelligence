```python
import pandas as pd
from sqlalchemy import create_engine, text

connection_string = f"postgresql+psycopg2://{'postgres'}:{2502}@{'localhost'}:{5432}/{'ecommerce_db'}"
engine = create_engine(connection_string)

# 2. Corrected the connection test syntax using text() for raw SQL execution
with engine.connect() as conn:
    result = conn.execute(text("SELECT current_date;"))
    print(result.fetchall())

# 3. Alternative: If you actually want to load data into a Pandas DataFrame
# df = pd.read_sql("SELECT * FROM your_table_name;", con=engine)
```

```python
df = pd.read_csv("../data/processed/online_retail_clean.csv", parse_dates= ["InvoiceDate"])
```

```python
# creating the table
create_table_sql = """
CREATE TABLE IF NOT EXISTS retail_transactions (
    invoice_no TEXT,
    stock_code TEXT,
    description TEXT,
    quantity INTEGER,
    invoice_date TIMESTAMP,
    unit_price NUMERIC(10, 2),
    customer_id TEXT,
    country TEXT,
    revenue NUMERIC(10, 2),
    is_cancelled BOOLEAN
);
"""


with engine.connect() as conn:
    conn.execute(text(create_table_sql)
    conn.commit()
```

```python
# 1. Select only the columns that exist in the SQL table
columns_to_keep = [
    'InvoiceNo', 'StockCode', 'Description', 'Quantity', 
    'InvoiceDate', 'UnitPrice', 'CustomerID', 'Country', 
    'Revenue', 'IsCancelled'
]
df_sql = df[columns_to_keep].copy()

# 2. Rename the columns to match the snake_case used in PostgreSQL
df_sql.columns = [
    'invoice_no', 'stock_code', 'description', 'quantity', 
    'invoice_date', 'unit_price', 'customer_id', 'country', 
    'revenue', 'is_cancelled'
]

# 3. Now send this new dataframe to SQL
df_sql.to_sql('retail_transactions', engine, if_exists='append', index=False)
```

```python
count = pd.read_sql("SELECT COUNT(*) FROM retail_transactions", engine)
print(count)
```
